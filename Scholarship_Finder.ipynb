```python

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from datetime import datetime

```

```python
# -------------------------------
# Scholarship Dataset
# -------------------------------
data = {
    "Scholarship Name": [
        "Merit Scholarship",
        "Minority Scholarship",
        "SC ST Scholarship",
        "Women in STEM Scholarship",
        "EWS Scholarship"
    ],
    "Description": [
        "Students with high marks above 80 percent",
        "Scholarship for minority students with low income",
        "Financial help for SC and ST category students",
        "Scholarship for female students in science and engineering",
        "Scholarship for economically weaker section students"
    ],
    "Min Marks": [80, 50, 50, 60, 55],
    "Category": ["any", "minority", "sc/st", "any", "any"],
    "Gender": ["any", "any", "any", "female", "any"],
    "Income Limit": [500000, 300000, 400000, 600000, 250000],
    "Deadline": [
        "30-04-2025",
        "15-05-2025",
        "20-05-2025",
        "10-06-2025",
        "25-04-2025"
    ]
}

df = pd.DataFrame(data)
df["Deadline"] = pd.to_datetime(df["Deadline"], format="%d-%m-%Y")

# -------------------------------
# Student Input
# -------------------------------
print("Enter Student Details")
marks = int(input("Marks Percentage: "))
category = input("Category (general/sc/st/minority): ").lower()
gender = input("Gender (male/female): ").lower()
income = int(input("Annual Family Income (INR): "))

student_profile = f"{category} {gender} student with {marks} percent marks and income {income}"

# -------------------------------
# NLP Similarity
# -------------------------------
vectorizer = TfidfVectorizer()
scholarship_vectors = vectorizer.fit_transform(df["Description"])
student_vector = vectorizer.transform([student_profile])

df["Similarity"] = cosine_similarity(student_vector, scholarship_vectors).flatten()

# -------------------------------
# Eligibility Filtering
# -------------------------------
eligible = df[
    (marks >= df["Min Marks"]) &
    (income <= df["Income Limit"]) &
    ((df["Category"] == "any") | (df["Category"] == category)) &
    ((df["Gender"] == "any") | (df["Gender"] == gender))
]

eligible = eligible.sort_values(by="Similarity", ascending=False)

# -------------------------------
# Deadline Reminder System
# -------------------------------
today = datetime.today()

print("\n🎓 Eligible Scholarships & Deadlines:\n")

if eligible.empty:
    print(" No scholarships found for your profile.")
else:
    for _, row in eligible.iterrows():
        days_left = (row["Deadline"] - today).days

        print(f" {row['Scholarship Name']}")
        print(f"    Deadline: {row['Deadline'].strftime('%d-%m-%Y')}")

        if days_left <= 0:
            print("    Deadline Passed\n")
        elif days_left <= 15:
            print(f"    Hurry! Only {days_left} days left\n")
        else:
            print(f"   {days_left} days remaining\n")

# -------------------------------
# End of Program
# -------------------------------
```
